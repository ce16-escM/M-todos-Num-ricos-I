Importando librerias importantes

```python
import math
import numpy as np
import pandas as pd  #Se usa para las tablas
import matplotlib.pyplot as plt
```

Queremos resolver la ecuación  $cos(x)-x=0$ hallando asi la raiz de la ecuación. La funcion asociada es $f(x)=cos(x)-x$

```python
#DECLARANDO LA FUNCION RESPECTIVA Y HACIENDO EL BOSQUEJO DE SU GRAFICA

def f(x):
  r=np.cos(x)-x
  return r

xx=np.linspace(-1,5,100)

plt.plot(xx,f(xx))
plt.axhline(y=0,color="black",label="Eje x")
plt.axvline(x=0,color="black",label="Eje y")

plt.title("Grafica f(x)=cos(x)-x")

```

Veamos que la funcion f(x) corta con y=0 en un valor poco menor a 1, veamos que resulta al evaluar $f(0.6)$ para ver que tan alejado esta de 0

```python
print(f" Evaluando una aproximacion de p, es decir p0=0.6  f(p0)={f(0.6)}")
```

Ahora calculamos la derivada de $f(x)=cos(x)-x$,es decir $f'(x)=-sen(x)-1$, decidi evaluar  $f'(p0)$, y resulto que es distinto de 0, entonces efectivamente podemos aplicar el metodo de newton

Sea $g(x)$ la funcion que representa la derivada

```python
def g(x):
  r2=-1*np.sin(x)-1
  return r2

Pendiente_EN_p0=g(0.6)
print(Pendiente_EN_p0)
```

```python
def newton(p0,tol):
  #tol=0.00000001  esta comentada esta parte porque el usuario va a dar el valor inicial de la aproximacion y la tolerancia al mandar llamar la funcion
  error=100
  Lerror=[]
  Lp_i=[p0]

  i=0

  print(f"n            /    pn-1     /       pn       /       error= |pn-pn-1|")

  while error>tol:
    p1=p0-(np.cos(p0)-p0)/(-1*np.sin(p0)-1)
    error=np.abs(p1-p0)
    Lerror.append(error)
    i=i+1

    print("{0} \t\t {1:6.6f} \t {2:6.6f} \t {3:6.6f} \t ".format(i,p0, p1,error ))

    Lp_i.append(p1)
    p0=p1
  return Lp_i,Lerror
```

```python
Resultados=newton(0.6,0.0001)
```
