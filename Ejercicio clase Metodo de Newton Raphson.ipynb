```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def f(x):
  return 3*x**2-np.exp(x)

xx=np.linspace(0,1,20)
plt.plot(xx,f(xx))
plt.axhline(y=0,color="red")
```

```python

tol=0.0001
error=100
p0=0.5

Lerror=[]
Lp_i=[]
while error>tol:
    p1=p0-(3*p0**2-np.exp(p0))//(6*p0-np.exp(p0))
    error=np.abs(p1-p0)
    Lerror.append(error)
    Lp_i.append(p1)
    p0=p1

#print("Lerror")
```

```python
def newton(p0,tol):
  #tol=0.00000001
  error=100
  Lerror=[]
  Lp_i=[p0]
  while error>tol:
    p1=p0-(3*p0**2-np.exp(p0))/(6*p0-np.exp(p0))
    error=np.abs(p1-p0)
    Lerror.append(error)
    Lp_i.append(p1)
    p0=p1
  return Lp_i,Lerror
```
