# **Método de Neville y resolución de un Ejercicio**

Importando librerias importantes

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

Se aplicara el método de Neville para lograr encontrar el polinomio de interpolación en un determinado intervalo

1ero necesitaremos los $n+1$ nodos(x0,x1,...,xn) y sus respectivas evaluaciones o $f(xi)$ para $0<=i<=n$

```python
#Datos de los n+1 nodos arbitrarios
X = [1.0, 2,3,4,5]
Y= [1,4,9,16,25]
```

Se crea una función especifica que calcule los polinomios de grado 0, grado 1, grado 2 y así sucesivamente

```python
def metodo_neville(X,Y, x_evaluar=None):
    n = len(X)

    if x_evaluar is None:
        x = sp.Symbol('x')
        Q = sp.zeros(n, n) # Matriz para representar la tabla de los polinomios calculados
    else:
        x = x_evaluar
        Q = np.zeros((n, n)) # Matriz numérica

    for i in range(n):
        Q[i, 0] = Y[i]

    for j in range(1, n):

        for i in range(j, n):
            # Fórmula de Neville
            term1 = (x - X[i-j]) * Q[i, j-1]
            term2 = (x - X[i]) * Q[i-1, j-1]
            denominador = X[i] - X[i-j]

            Q[i, j] = (term1 - term2) / denominador

    polinomio_final = Q[n-1, n-1]

    return polinomio_final, Q


print("--- Calculando el polinomio simbólico ---")
polinomio, tabla_Q = metodo_neville(X, Y)


polinomio_simplificado = sp.simplify(sp.expand(polinomio))
polinomio_redondeado = sp.N(polinomio_simplificado, 6)

print("El polinomio P(x) que pasa por los nodos dados es:")
print(f"P(x) = {polinomio_redondeado}")


```

```python
#Declaramos la informacion de los nodos(puntos)
X1=[1.0, 1.3, 1.6, 1.9, 2.2]

Y1=[0.7651977, 0.6200860, 0.4554022, 0.2818186, 0.1103623]
```

Si queremos evaluar la aproximación en un punto especifico tenemos que declarar la variable

```python
x_punto = 1.5
print(f"\n--- Evaluando en x = {x_punto} ---")
resultado_evaluado, _ = metodo_neville(X1,Y1, x_evaluar=x_punto)
print(f"P({x_punto}) = {resultado_evaluado:.8f}")

_, tabla_numerica = metodo_neville(X1, Y1, x_evaluar=x_punto)
n = len(X)

for j in range(n):
    print(f"Polinomios de grado {j}:")
    for i in range(j, n):
        # Imprimimos con formato para que queden alineados
        print(f"  Q_{i},{j} = {tabla_numerica[i, j]:.8f}")
    print("-" * 40)


```
