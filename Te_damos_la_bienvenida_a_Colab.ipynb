```
import numpy as np
import matplotlib.pyplot as plt
```

```
print("Vamos a definir la función f(x)= e^(x^2)")
```

```
 def f(x):
    return np.exp(x**2)
```

```
print(f(0))
```

```
xx=np.linspace(0,2,100)
plt.plot(xx,f(xx))
```

```
xx=np.linspace(-1.1,1.1,100)
plt.plot(xx,f(xx))
```
